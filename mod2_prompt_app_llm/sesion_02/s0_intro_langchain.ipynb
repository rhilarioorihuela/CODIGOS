### 4. Crear función para obtener el modelo LLM

```python
# Generar una funcion que obtenga el modelo LLM

from langchain.chat_models import init_chat_model

def get_llm(prefix_model, temperature=0.9):
    """
    Obtiene un modelo de lenguaje (LLM) basado en el modelo especificado.

    Args:
        model (str): El nombre del modelo a utilizar.
        temperature (float, optional): La temperatura para la generación de texto. 
                                        Valores más altos producen respuestas más creativas.
                                        Valores más bajos producen respuestas más deterministas.
                                        Por defecto es 0.9.

    Returns:
        llm: Un objeto de modelo de lenguaje inicializado.
    """
    llm = init_chat_model(
                            model=prefix_model, 
                            # temperature=temperature
                          )
    return llm
```

```python
# Configuración del modelo de lenguaje
prefix_model = os.getenv("PREFIX_MODEL")
prompt = "¿Cuál es la capital de Perú?"

# Obtención del modelo
llm = get_llm(prefix_model)

# Uso del modelo de lenguaje
response = llm.invoke(prompt)

# Imprime la respuesta del modelo
print(response.text)
```

### 6. Stream

```python

load_dotenv()
prefix_model = os.getenv("PREFIX_MODEL")


from langchain_core.messages import SystemMessage, HumanMessage

msgs =  [
        # Mensajes del sistema 
        SystemMessage(content=
                    """
                        Eres un asistente experto en inteligencia artificial. 
                        Responde de forma breve y precisa, usando como máximo 100 palabras.
                    """),
        # Mensajes del usuario
        HumanMessage(content=
                        """
                        ¿Qué es un agente de inteligencia artificial?
                        """)

        ]

# Obtención del modelo
llm = get_llm(prefix_model)

# Uso del modelo de lenguaje
for chunk in llm.stream(msgs):
    print(chunk.text, end="", flush=True)
print()  # Para agregar un salto de línea después de la respuesta
```

### 8. Aplicando parseadores a las respuestas del modelo de IA

```python
# Generar una funcion que obtenga el modelo LLM

from langchain.chat_models import init_chat_model

def get_llm(prefix_model, temperature=0.9):
    """
    Obtiene un modelo de lenguaje (LLM) basado en el modelo especificado.

    Args:
        model (str): El nombre del modelo a utilizar.
        temperature (float, optional): La temperatura para la generación de texto. 
                                        Valores más altos producen respuestas más creativas.
                                        Valores más bajos producen respuestas más deterministas.
                                        Por defecto es 0.9.

    Returns:
        llm: Un objeto de modelo de lenguaje inicializado.
    """
    llm = init_chat_model(
                            model=prefix_model, 
                            # temperature=temperature
                          )
    return llm
```

```python
from langchain_core.prompts import PromptTemplate
import os
from dotenv import load_dotenv

load_dotenv()

prefix_model = os.getenv("PREFIX_MODEL")

# print(prefix_model)

# Generar una funcion que obtenga el modelo LLM
TEMPLATE = """
               Saluda a {nombre} en el idioma {idioma}.    
            """

saludo_template = PromptTemplate.from_template(TEMPLATE)

# Se obteniene el prompt con los valores de nombre y idioma
prompt = saludo_template.invoke({"nombre": "Jaime", 
                                 "idioma": "español"})

# Obtención del modelo
llm = get_llm(prefix_model)

# Uso del modelo de lenguaje
response = llm.invoke(prompt)

#print(response)

# Aplicar parseadores de salida (si es necesario)
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

parser_response = parser.invoke(response)

# Imprime la respuesta del modelo
print(parser_response)

```

### 9. LCEL : LangChain Expression Language

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
import os
from dotenv import load_dotenv

load_dotenv()

prefix_model = os.getenv("PREFIX_MODEL")

#######################
# Definicion de capas de abstracción para el prompt y el modelo de lenguaje
#######################

# Set template
TEMPLATE = """
               Saluda a {nombre} en el idioma {idioma}.    
            """
prompt_template = PromptTemplate.from_template(TEMPLATE)

# Set LLM
llm    = get_llm(prefix_model)

# Set output parser
parser = StrOutputParser()


# CASO 1 : Ejecucion del flujo de trabajo: prompt -> modelo -> parseador
# Ejecucion del flujo de trabajo: prompt -> modelo -> parseador
prompt          = prompt_template.invoke({"nombre": "Jaime", 
                                          "idioma": "Ingles"})
response        =             llm.invoke(prompt)
parser_response =          parser.invoke(response)

# Imprime la respuesta del modelo
print(f"CASO 1 - Respuesta del modelo: {parser_response}")

```

```python

# CASO 2 : Ejecución con LCEL

chain = prompt_template | llm | parser

parser_response = chain.invoke({"nombre": "Jaime", 
                                "idioma": "Ingles"})

# Imprime la respuesta del modelo
print(f"CASO 2 - Respuesta del modelo: {parser_response}")
```

```python
from langchain_core.messages import SystemMessage, HumanMessage

# Usando memoria

historial = [SystemMessage(content="""
                            Eres un asistente experto. Responde de forma breve y precisa, 
                            usando como máximo 20 palabras.
                    """)]

def my_invoke(prompt):

    historial.append(HumanMessage(content=prompt))
    respuesta = llm.invoke(historial)
    #print(type(respuesta))
    historial.append(respuesta)

    return respuesta
```

```python
print("Salida :" + my_invoke("Hola, me llamo Jaime").text)
```

```python
print("Salida : " + my_invoke("Hola, ¿Como me llamo?").text)
```
